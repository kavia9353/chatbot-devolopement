
# Week 9 – Chatbot Development

## Building an Interactive AI Chatbot using Ollama

### Objective
- Understand the architecture of an AI chatbot powered by a local Large Language Model.
- Build a command-line chatbot using Python and Ollama.
- Implement conversation history for context-aware responses.
- Prepare the chatbot for future integration with LangChain.

### Domain
Education and Student Support

### Technologies Used
- Python
- Ollama
- Llama 3.2
- Pandas
- Jupyter Notebook
- Visual Studio Code

## 1. Project Description

This project develops an interactive Education AI Chatbot using Python and the Ollama platform.

The chatbot uses the locally installed Llama 3.2 language model to answer student questions.

It supports continuous conversations, maintains conversation history, and saves user questions and chatbot responses in CSV files with timestamps.

The chatbot helps students with Python, SQL, Artificial Intelligence, examination preparation, study planning, and communication skills.


## 2. Observations

1. The chatbot generates responses using the locally running Llama 3.2 model.
2. It accepts multiple questions through a continuous conversation loop.
3. Conversation history helps the chatbot maintain context between questions.
4. User questions and chatbot responses are stored with timestamps.
5. The chatbot can assist students with programming, AI, SQL, and study-related questions.

## 3. Strengths

- Uses a locally running language model.
- Supports continuous conversations.
- Maintains conversation context.
- Saves conversations for future analysis.
- Provides assistance with education-related questions.

## 4. Limitations

- Responses may sometimes be incorrect or incomplete.
- Response quality depends on the language model.
- The chatbot requires Ollama to be running.
- It is currently a command-line application.
- It does not guarantee that every answer is factually correct.

## 5. Conclusion

This project helped demonstrate how to build an interactive AI chatbot using Python and Ollama. Conversation history and CSV storage were implemented to support context-aware interactions and future analysis. The project provides a foundation for integrating the chatbot with LangChain in the next week.
